Split the data into input feature and output feature


```python
## X is the input feature
X = dataframe.drop(columns=['Outcome'], axis=1)
## y is the output feature
y = dataframe['Outcome']
```

```python
X
```

Data Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
x_scaled = scaler.fit_transform(X)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (15,15))
sns.boxplot(data=x_scaled)
plt.savefig('boxplot.png')
```

```python
from scipy import stats

z_scores = np.abs(stats.zscore(x_scaled))
print("Z-scores:\n", z_scores)

# Threshold for identifying an outlier (usually 3)
threshold = 3

# Filtering out the outliers
df_no_outliers = x_scaled[(z_scores < threshold).all(axis=1)]

print("\nData without outliers:\n", df_no_outliers)
```

```python
import pandas as pd

# Convert the NumPy array to a Pandas DataFrame
x_scaled_df = pd.DataFrame(x_scaled)
```

```python
Q1 = x_scaled_df.quantile(0.25)
Q3 = x_scaled_df.quantile(0.75)
IQR = Q3 - Q1
print(IQR)

```

```python
df_no_outliers = x_scaled_df[~((x_scaled_df < (Q1 - 1.5 * IQR)) | (x_scaled_df > (Q3 + 1.5 * IQR))).any(axis=1)]
print(df_no_outliers)
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (15,15))
sns.boxplot(data=df_no_outliers)
plt.savefig('boxplot.png')
```
